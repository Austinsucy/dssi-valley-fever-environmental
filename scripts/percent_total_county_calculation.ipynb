```python
counts_by_county = pd.read_excel('/content/drive/MyDrive/DSSI/Project/data/Cocci Case Counts by County and Month, 2001-2024.xlsx', header=1)
counts_by_county = counts_by_county.drop(counts_by_county.columns[[3,4]], axis=1)
counts_by_county.columns = ['County','Year-Month','Case_Count']
counts_by_county['Case_Count'] = pd.to_numeric(counts_by_county['Case_Count'], errors='coerce')
counts_by_county['Case_Count'] = counts_by_county['Case_Count'].fillna(0)
counts_by_county['County'] = counts_by_county['County'].str.title()
counts_by_county
```

```python
county_totals = counts_by_county.groupby('County')['Case_Count'].sum().reset_index()
county_totals
```

```python
county_totals.sort_values(by='Case_Count',ascending=False)
```

```python
total_all_counties = counts_by_county['Case_Count'].sum()
total_all_counties
```

```python
selected_counties = [
    'Fresno',
    'Kern',
    'Los Angeles',
    'Tulare',
    'San Diego',
    'San Luis Obispo',
    'Orange',
    'Ventura'
]

selected_county_total = counts_by_county[counts_by_county['County'].isin(selected_counties)]['Case_Count'].sum()
selected_county_total
```

```python
selected_out_of_total = selected_county_total/total_all_counties
selected_out_of_total
```
